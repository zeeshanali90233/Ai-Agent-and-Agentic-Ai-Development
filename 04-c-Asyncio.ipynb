# Python Concurrency

Event Loop, Async, Multithreading, and Multiprocessing

## 1. Event Loop
The asyncio event loop schedules tasks that are waiting for I/O and switches to other ready tasks instead of blocking.

```python
import asyncio

async def task(name, delay):
    print(f"{name} started")
    await asyncio.sleep(delay)
    print(f"{name} finished")

async def main():
    await asyncio.gather(task("Task 1",3), task("Task 2",2))

await main()
```

## 2. Async / Await
`async` defines a coroutine. `await` pauses the coroutine without blocking the thread.

```python
import asyncio

async def download():
    print("Downloading...")
    await asyncio.sleep(2)
    print("Done")

await download()
```

## 3. Multithreading
Threads are useful for I/O-bound work. In CPython, the GIL allows only one thread to execute Python bytecode at a time.

```python
import threading, time

def work():
    print("Working...")
    time.sleep(2)
    print("Finished")

t1=threading.Thread(target=work)
t2=threading.Thread(target=work)
t1.start(); t2.start()
t1.join(); t2.join()
```

## 4. Multiprocessing
Processes bypass the GIL and are best for CPU-bound work.

```python
from multiprocessing import Process

def work():
    print("Working in process")

if __name__ == "__main__":
    p1 = Process(target=work)
    p2 = Process(target=work)

    p1.start()
    p2.start()

    p1.join()
    p2.join()
```

## Comparison

| Feature | Asyncio | Threading | Multiprocessing |
|---|---|---|---|
| Best for | I/O | I/O | CPU |
| Event Loop | Yes | No | No |
| Multiple CPU cores | No | Limited by GIL | Yes |
| Memory | Low | Medium | High |

### Rule of thumb
- APIs, databases, networking → Asyncio
- Blocking I/O libraries → Threads
- Heavy computation / AI / image processing → Multiprocessing
